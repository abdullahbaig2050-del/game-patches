# Question no : 19

```python
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    else:
        print("Age valid hai, entry allowed!")

# Test
try:
    check_age(16)
except ValueError as e:
    print(f"Error: {e}")

check_age(20)
```
