### **Question 30**
Dataset ko poora clean karne ke baad safely ek nayi csv file 'cleaned_movies.csv' ke naam se save karein.

```python
clean_df = df.copy()

# 1) Duplicate rows hatao
clean_df = clean_df.drop_duplicates()

# 2) Released_Year ki invalid value (NaN) wali row hatao aur int bana do
clean_df = clean_df.dropna(subset=["Released_Year"])
clean_df["Released_Year"] = clean_df["Released_Year"].astype(int)

# 3) Missing Certificate ko "Unrated" aur Meta_score ko median se fill karo
clean_df["Certificate"] = clean_df["Certificate"].fillna("Unrated")
clean_df["Meta_score"] = clean_df["Meta_score"].fillna(clean_df["Meta_score"].median())

# (Runtime numeric aur Gross median-filled pehle hi ho chuka hai)
print("Remaining missing values:", clean_df.isnull().sum().sum())

clean_df.to_csv("cleaned_movies.csv", index=False)
print("cleaned_movies.csv save ho gayi, shape:", clean_df.shape)
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
drama_g = df[df["Genre"].str.contains("Drama", na=False)].assign(Genre_Group="Drama")
action_g = df[df["Genre"].str.contains("Action", na=False)].assign(Genre_Group="Action")
compare = pd.concat([drama_g, action_g])

plt.figure(figsize=(9, 6))
sns.violinplot(data=compare, x="Genre_Group", y="Gross", hue="Genre_Group",
               palette="Set2", inner="box", legend=False)
plt.title("Gross Revenue: Drama vs Action", fontsize=14, fontweight="bold")
plt.xlabel("Genre")
plt.ylabel("Gross Revenue")
plt.tight_layout()
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = df.sort_values("IMDB_Rating", ascending=False).head(5).copy()
top5["IMDB_x10"] = top5["IMDB_Rating"] * 10

x = np.arange(len(top5))
w = 0.38

plt.figure(figsize=(11, 6))
plt.bar(x - w/2, top5["Meta_score"], w, label="Meta Score", color="#4C72B0")
plt.bar(x + w/2, top5["IMDB_x10"], w, label="IMDB Rating (x10)", color="#DD8452")
plt.xticks(x, top5["Series_Title"], rotation=20, ha="right")
plt.ylabel("Score (out of 100)")
plt.title("Top 5 Rated Movies: Meta Score vs IMDB Rating (100 scale)", fontsize=14, fontweight="bold")
plt.legend()
plt.tight_layout()
plt.show()
```
